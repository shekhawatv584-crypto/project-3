## Feature Engineering

```python
# ============================================
#  FEATURE ENGINEERING
#  CHECK PRICE VARIABLES
# ============================================

df[["product_price", "quantity", "discount_percentage", "final_price"]].head(10)
```

```python
# ============================================
#  CREATE DISCOUNT CATEGORY
# ============================================

df["discount_category"] = pd.cut(
    df["discount_percentage"],
    bins=[-1, 10, 20, 30],
    labels=["Low Discount", "Medium Discount", "High Discount"]
)

df["discount_category"].value_counts().sort_index()
```

```python
# ============================================
#  CREATE DELIVERY CATEGORY
# ============================================

df["delivery_category"] = pd.cut(
    df["delivery_days"],
    bins=[0, 3, 7, 100],
    labels=["Fast", "Standard", "Slow"]
)

df["delivery_category"].value_counts().sort_index()
```

```python
# CREATE AGE GROUP

df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 18, 30, 45, 60, 100],
    labels=["Under 18", "18-30", "31-45", "46-60", "60+"]
)

print(df["age_group"].value_counts().sort_index())
```

```python
# IQR OUTLIER DETECTION FOR FINAL PRICE

Q1 = df["final_price"].quantile(0.25)
Q3 = df["final_price"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["final_price"] < lower_bound) |
    (df["final_price"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", len(outliers))
```

```python
# OUTLIER PERCENTAGE

outlier_percentage = (len(outliers) / len(df)) * 100

print("Outlier Percentage:", round(outlier_percentage, 2), "%")
```

```python
# HIGH-VALUE TRANSACTION ANALYSIS

high_value_orders = df[df["final_price"] > upper_bound]

print("High-value transactions:", len(high_value_orders))
print("\nAverage Quantity:")
print(high_value_orders["quantity"].mean())

print("\nAverage Product Price:")
print(high_value_orders["product_price"].mean())

print("\nProduct Category:")
print(high_value_orders["product_category"].value_counts())

print("\nCustomer Segment:")
print(high_value_orders["customer_segment"].value_counts())
```

```python
# Outlier Summary
total_records = len(df)
outlier_count = 1335
normal_count = total_records - outlier_count

outlier_percentage = (outlier_count / total_records) * 100
normal_percentage = (normal_count / total_records) * 100

print("Total Records:", total_records)
print("Normal Records:", normal_count)
print("Outlier Records:", outlier_count)
print("Normal Percentage:", round(normal_percentage, 2), "%")
print("Outlier Percentage:", round(outlier_percentage, 2), "%")
```

## OUTLIER STATISTICAL ANALYSIS

```python
# FINAL PRICE OUTLIER ANALYSIS

Q1 = df["final_price"].quantile(0.25)
Q3 = df["final_price"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["final_price"] < lower_bound) |
    (df["final_price"] > upper_bound)
]

plt.figure(figsize=(12, 7))

sns.boxplot(
    y=df["final_price"],
    color="steelblue"
)

plt.axhline(
    upper_bound,
    color="firebrick",
    linestyle="--",
    linewidth=2,
    label=f"Upper Bound = {upper_bound:,.0f}"
)

plt.title(
    "Final Price Outlier Analysis using IQR",
    fontsize=17,
    fontweight="bold"
)

plt.ylabel("Final Price")

plt.legend()
plt.grid(axis="y", alpha=0.2)

plt.tight_layout()
plt.show()

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", len(outliers))
print("Outlier Percentage:", round(len(outliers) / len(df) * 100, 2), "%")
```
